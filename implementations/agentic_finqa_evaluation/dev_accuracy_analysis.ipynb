```python
import sys
import uuid
from pathlib import Path
import pandas as pd
import json
import os
```

```python
dev_responses = []

for root, dirs, files in os.walk('./meps/gemini_gemini/chartqapro/test'):
     for file in files:
        with open(os.path.join(root, file), "r") as auto:
            mep = json.loads(auto.read())
            dev_responses.append({
                'id': mep['sample']['sample_id'],
                'question': mep['sample']['question'],
                'planner_answerable': mep['plan']['parsed']['answerability_check'],
                'analyst_answer': mep['analyst']['parsed']['answer'],
                'verifier_answer': mep['verifier']['parsed']['answer'],
                'verifier_correct': mep['verifier']['parsed']['verdict']
            })
```

```python
pd.DataFrame(dev_responses)
```

```python
df_dev = pd.read_json("./FinQA_dataset/dev.json")
df_dev['question'] = df_dev['qa'].apply(lambda x: x['question'])
df_dev['answer'] = df_dev['qa'].apply(lambda x: x['answer'])
```

```python
df_dev_meps = pd.merge(df_dev[['question', 'answer', 'id']], pd.DataFrame(dev_responses), on=['question', 'id'], how='inner')
```
